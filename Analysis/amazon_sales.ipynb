```python
print(df['category'].unique())
print(df['rating'].unique())
print(df['rating_count'].unique())
print(df['discount_percentage'].unique())
```

```python
drop_cols=['product_id','about_product','user_id','user_name','review_id','review_title','review_content','img_link','product_link']
df=df.drop(columns=drop_cols,errors='ignore')
```

```python
for col in df.columns:
    print(col)
```

```python
print(df['discounted_price'].head())
print(df['discount_percentage'].head())
```

```python
df['actual_price']=(
    df['actual_price']
    .str.replace('₹','',regex=False)
    .str.replace(',','',regex=False)
    .astype(float)
)

df['discounted_price']=(
    df['discounted_price']
    .str.replace('₹','',regex=False)
    .str.replace(',','',regex=False)
    .astype(float)
)

df['discount_percentage']=(
    df['discount_percentage'].str.replace('%','',regex=False)
    .astype(float)
)
```

```python
df['rating_count']=(
    df['rating_count']
    .str.replace(',','',regex=False)
    .astype(float)
)
```

```python
print(df['rating'].dtype)
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating'].dtype
```

```python
df['rating'].isna().sum()
```

```python
df[df['rating'].isna()]
```

```python
df.describe()
```

Key Observations:
-75% products under ₹2000,Majority of products are low-to-mid range.
-A few premium products significantly increase the average price.
-Most products have ratings above the average, indicating generally positive customer feedback.
-The average discount percentage is around 50%, with a wide range between minimum and maximum discounts, indicating varied pricing strategies.
-Rating count is highly skewed, where a small number of products receive very high engagement compared to the majority.
-Product prices vary widely, with a few high-priced items compared to many low-priced products.
-Price distribution is right-skewed due to a few high-priced products.
-Product ratings are generally high and show low variability.
-Heavy discounting is not rare — it’s a common pricing strategy.

```python
plt.figure(figsize=(14, 6))
plt.hist(df[df['discounted_price'] <= 5000]['discounted_price'], bins=20)
plt.xticks(np.arange(0, 5001, 500))
plt.title("Distribution of Discounted Prices", fontsize=14)
plt.xlabel("Discounted Price", fontsize=12)
plt.ylabel("Number of Products", fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.hist(df['discount_percentage'], bins=20)
plt.xticks(np.arange(0, 101, 10))

plt.title("Distribution of Discount Percentage")
plt.xlabel("Discount Percentage")
plt.ylabel("Number of Products")

plt.show()
```

```python
category_summary = df.groupby('category').agg({
    'discounted_price': 'mean',
    'discount_percentage': 'mean',
    'rating': 'mean',
    'rating_count': 'mean'
}).sort_values(by='discounted_price', ascending=False)

category_summary
```

```python
top_categories = category_summary.head(10)
top_categories
```

```python
plt.figure(figsize=(12, 6))
plt.bar(top_categories.index, top_categories['discounted_price'])
plt.xticks(rotation=45, ha='right')
plt.title("Top 10 Categories by Average Discounted Price")
plt.xlabel("Category")
plt.ylabel("Average Discounted Price")
plt.show()
```

Key Obseravtions:
-Air Conditioners and Laptops have the highest average discounted prices, indicating that premium home appliances and computing devices dominate the high-price segment even after discounts.
-External Solid State Drives (SSDs) stand out by offering very high average discounts (~68%) despite being mid-range priced products, suggesting aggressive promotional strategies in storage devices.
-Smartphones, although relatively expensive, have lower average discounts (~23%), indicating strong demand and less reliance on heavy discounting.
-High-priced categories consistently maintain strong ratings (above 4.0), showing that customers are generally satisfied with premium products despite higher prices.
-Customer engagement varies significantly across categories — Smartphones and SSDs have very high rating counts, while categories like laptops show comparatively lower engagement, likely due to less frequent purchases.

```python
top_discount_categories = category_summary.sort_values(
    by='discount_percentage',
    ascending=False
).head(10)

top_discount_categories
```

```python
plt.figure(figsize=(12, 6))
plt.bar(top_discount_categories.index,
        top_discount_categories['discount_percentage'])
plt.xticks(rotation=45, ha='right')
plt.title("Top 10 Categories by Average Discount Percentage")
plt.xlabel("Category")
plt.ylabel("Average Discount Percentage")
plt.show()
```

Key Observation:
-Low-cost accessory categories such as phone charms, ear pads, cables, and covers receive extremely high average discounts (above 75%), indicating aggressive promotional tactics to drive volume sales.
-Despite heavy discounting, these categories maintain moderate to high ratings (around 4.0), suggesting acceptable perceived quality even at reduced prices.
-High rating counts for accessories like phone stands and cases indicate strong customer engagement, likely due to frequent purchases and impulse buying behavior.
